```python
from pathlib import Path
import pandas as pd

source = Path(
    r"C:\Users\imran\Desktop\Capstone raw data\RNAseq_capstone"
    r"\data\raw\All_results_FPKM_Annot.xlsx"
)
output = source.parent / "annotation_core_raw.csv"

df = pd.read_excel(source, sheet_name="JoinDataAll-4", dtype="string")
df.columns = [str(column).strip() for column in df.columns]

# Keep the fields needed for gene descriptions and pathway analysis.
columns_to_keep = [
    "Gene_ID",
    "gene_name",
    "gene_description",
    "tf_family",
    "gene_biotype",
    "kegg_pathway_name",
    "kegg_pathway_gene_id",
    "kegg_pathway_id",
    "KOG Description",
    "KOG ID",
]

missing = [column for column in columns_to_keep if column not in df.columns]
if missing:
    raise ValueError(
        f"These columns were not found: {missing}\n"
        f"Available columns: {list(df.columns)}"
    )

small = df[columns_to_keep].copy()

# Keep repeated rows: they may contain separate gene-to-pathway links.
small.to_csv(output, index=False, encoding="utf-8-sig")

print(f"Rows exported: {len(small):,}")
print(f"Distinct nonblank genes: {small['Gene_ID'].nunique():,}")
print(f"Saved: {output}")
```
