```python
import csv

input_file = r"C:\Users\imran\Desktop\RNAseq_capstone\data\processed\annotation.csv"
output_file = r"C:\Users\imran\Desktop\RNAseq_capstone\data\processed\annotation_clean.csv"

expected_columns = 12   # your CSV has 12 columns

bad_rows = []
clean_rows = []

with open(input_file, "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter=",")
    for i, row in enumerate(reader):
        if len(row) != expected_columns:
            bad_rows.append((i+1, row))
            
            # Fix the row by padding missing columns
            if len(row) < expected_columns:
                row += [""] * (expected_columns - len(row))
            else:
                row = row[:expected_columns]
        
        clean_rows.append(row)

print("Bad rows found:")
for line_number, row in bad_rows:
    print(f"Line {line_number}: {row}")

with open(output_file, "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(clean_rows)

print("\nClean file saved as:", output_file)
```
